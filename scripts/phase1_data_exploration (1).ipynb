```python
from datasets import load_dataset


class DataLoader:

    DATASET_NAME = "KFUPM-JRCAI/arabic-generated-abstracts"

    def __init__(self):
        self.dataset = None

    def load_data(self):

        print("Loading dataset...")

        self.dataset = load_dataset(self.DATASET_NAME)

        print("Dataset loaded successfully!")

        return self.dataset
```

```python
loader = DataLoader()
```

```python
dataset = loader.load_data()
```

```python
class DataExplorer:

    def __init__(self, data):
        self.df = data.to_pandas() if hasattr(data, "to_pandas") else data.copy()

    def show_columns(self):
        print("\nColumns:")
        print(self.df.columns.tolist())

    def show_data_types(self):
        print("\nData Types:")
        print(self.df.dtypes)

    def check_missing_values(self):
        print("\nMissing Values:")
        print(self.df.isnull().sum())

    def check_duplicates(self):
        print("\nDuplicate Rows:")
        print(self.df.duplicated().sum())

    def check_empty_values(self):
        print("\nEmpty Values:")

        for column in self.df.columns:
            empty_values = (
                self.df[column]
                .fillna("")
                .astype(str)
                .str.strip()
                .eq("")
                .sum()
            )

            print(column, "->", empty_values)

    def check_inconsistencies(self):
        print("\nInconsistencies:")

        for column in self.df.columns:

            non_string_values = (
                ~self.df[column].dropna()
                .apply(lambda x: isinstance(x, str))
            ).sum()

            print(
                column,
                "-> non-string values:",
                non_string_values
            )

    def show_shape(self):
        print("\nDataset Shape:")
        print("Rows:", len(self.df))
        print("Columns:", len(self.df.columns))
```

```python
explorer_original = DataExplorer(
    dataset["from_title_and_content"]
)

explorer_original.show_columns()
explorer_original.show_data_types()
explorer_original.check_missing_values()
explorer_original.check_duplicates()
explorer_original.check_empty_values()
explorer_original.check_inconsistencies()
explorer_original.show_shape()
```

```python
import pandas as pd

df = dataset["from_title_and_content"].to_pandas()

human_data = pd.DataFrame({
    "text": df["original_abstract"],
    "label": "human"
})

ai_data = pd.concat([
    pd.DataFrame({
        "text": df["allam_generated_abstract"],
        "label": "AI-generated"
    }),
    pd.DataFrame({
        "text": df["jais_generated_abstract"],
        "label": "AI-generated"
    }),
    pd.DataFrame({
        "text": df["llama_generated_abstract"],
        "label": "AI-generated"
    }),
    pd.DataFrame({
        "text": df["openai_generated_abstract"],
        "label": "AI-generated"
    })
], ignore_index=True)

labeled_df = pd.concat(
    [human_data, ai_data],
    ignore_index=True
)

print("Target Distribution:")
print(labeled_df["label"].value_counts())
```

```python
labeled_df.to_csv(
    "from_title_and_content_labeled.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Labeled dataset saved successfully!")
```
