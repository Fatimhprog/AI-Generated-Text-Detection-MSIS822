```python
import os
os.makedirs("data/raw", exist_ok=True)
```

```python
from datasets import load_dataset

dataset = load_dataset("KFUPM-JRCAI/arabic-generated-abstracts")
```

```python
splits = ["by_polishing", "from_title", "from_title_and_content"]

for split in splits:
    df = dataset[split].to_pandas()
    df.to_csv(f"data/raw/{split}.csv", index=False)
    print(f"تم حفظ ملف: data/raw/{split}.csv")
```

```python
import pandas as pd

df_polish = pd.read_csv("data/raw/by_polishing.csv")
df_title = pd.read_csv("data/raw/from_title.csv")
df_full = pd.read_csv("data/raw/from_title_and_content.csv")
```

```python

human_rows = [{"text": row["original_abstract"], "label": "human"} for _, row in df.iterrows()]

ai_rows = []
for _, row in df.iterrows():
    ai_rows.append({"text": row["allam_generated_abstract"], "label": "ai_generated"})
    ai_rows.append({"text": row["jais_generated_abstract"], "label": "ai_generated"})
    ai_rows.append({"text": row["llama_generated_abstract"], "label": "ai_generated"})
    ai_rows.append({"text": row["openai_generated_abstract"], "label": "ai_generated"})

df_labels = pd.DataFrame(human_rows + ai_rows)
```
